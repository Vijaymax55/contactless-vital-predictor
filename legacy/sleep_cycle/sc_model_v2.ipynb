```python
import os 
import pandas as pd
import numpy as np
import io
from ast import literal_eval
from scipy.interpolate import PchipInterpolator
from scipy.signal import butter, lfilter, hilbert
from sklearn.model_selection import train_test_split
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score, classification_report
```

```python
folder_path = '/Users/vijayprakash/Documents/sc_model_jan/5_jan_mat'
csv_files = [f for f in os.listdir(folder_path) if f.endswith('.csv')]
merged_data = pd.DataFrame()

for csv_file in csv_files:
    file_date = csv_file.split('_')[2].split('.')[0]
    df = pd.read_csv(os.path.join(folder_path, csv_file), encoding='utf-8')
    df['Date'] = pd.to_datetime(file_date + ' ' + df['Timestamp'])
    df = df.drop(['Timestamp'], axis=1)
    merged_data = pd.concat([merged_data, df], ignore_index=True, sort=False)

merged_data = merged_data.sort_values(by=['Date'], ignore_index=True)
merged_data = merged_data.rename(columns={'Date': 'Time'})
merged_data['Time'] = merged_data['Time'].astype(str).str[:-6]
merged_data
```

```python
merged_data['Time'] = pd.to_datetime(merged_data['Time'], format='%Y-%m-%d %H:%M:%S')

# Round timestamps to the nearest minute and convert to string
merged_data['Time'] = merged_data['Time'].dt.floor('S').dt.strftime('%Y-%m-%d %H:%M:%S')

# Group by the rounded timestamp and aggregate the data
merged_data = merged_data.groupby('Time').agg(lambda x: x.tolist()).reset_index()
merged_data.to_csv('merged_data_5_jan.csv')
merged_data
```

```python

fs = 490
lowcut = 0.5
highcut = 6.0

def interpolate_to_490(arr):
    if len(arr) < 2:
        return None

    pchip_interpolator = PchipInterpolator(np.arange(len(arr)), arr)
    x_interp = np.linspace(0, len(arr) - 1, 490)
    y_interp = pchip_interpolator(x_interp)
    return y_interp.tolist()

def butter_bandpass(lowcut, highcut, fs, order=3):
    nyq = 0.5 * fs
    low = lowcut / nyq
    high = highcut / nyq
    b, a = butter(order, [low, high], btype='band')
    return b, a

def butter_bandpass_filter(data, lowcut, highcut, fs, order=3):
    b, a = butter_bandpass(lowcut, highcut, fs, order=order)
    y = lfilter(b, a, data)
    return y

# Read data from "merged_data.csv" and convert string representations to lists
merged_data = pd.read_csv('merged_data_5_jan.csv', converters={'Channel1': literal_eval, 'Channel2': literal_eval, 'Channel3': literal_eval, 'Channel4': literal_eval})

# Example usage
channels = ['Channel1', 'Channel2', 'Channel3', 'Channel4']

# Copy the DataFrame to keep the original data intact
df_interpolated = merged_data.copy()

# Apply interpolation only to the specified channels
df_interpolated[channels] = df_interpolated[channels].applymap(interpolate_to_490)

# Apply Butterworth filter to the specified channels
for channel in channels:
    df_interpolated[channel] = df_interpolated[channel].apply(lambda x: butter_bandpass_filter(x, lowcut, highcut, fs))

# Add the 'Time' column from the original DataFrame to the final DataFrame
df_interpolated['Time'] = merged_data['Time']

df_interpolated
```

```python
sleep_data = pd.read_csv('/Users/vijayprakash/Documents/sc_model_v2/5_jan_sleep/Sleep Analysis-2024-01-05-2024-01-05.csv')

sleep_data = sleep_data[(sleep_data['Source'] != 'Vijay’s iPhone') & (sleep_data['Value'] != 'InBed')] # Filter sleep_data
sleep_data = sleep_data.drop(['Source', 'Duration (hr)'], axis=1)
sleep_data['Value'] = sleep_data['Value'].replace('Core', 'LS')

sleep_data['Start'] = pd.to_datetime(sleep_data['Start']) # Convert 'Start' and 'End' columns to datetime
sleep_data['End'] = pd.to_datetime(sleep_data['End'])

# Expand time ranges and create a new DataFrame
expanded_data = pd.concat([pd.DataFrame({'Time': pd.date_range(row['Start'], row['End'], freq='S'), 'Value': row['Value']}) for _, row in sleep_data.iterrows()], ignore_index=True)

# Add a 'Time' column with both date and time
expanded_data['Time'] = pd.to_datetime(expanded_data['Time'].dt.strftime('%Y-%m-%d %H:%M:%S'))

df_interpolated['Time'] = pd.to_datetime(df_interpolated['Time']) # Convert 'Time' columns to datetime
expanded_data['Time'] = pd.to_datetime(expanded_data['Time'])

df_interpolated = df_interpolated.sort_values(by='Time') # Sort both DataFrames by 'Time' column
expanded_data = expanded_data.sort_values(by='Time')

master_data = pd.merge_asof(df_interpolated, expanded_data, on='Time') # Merge datasets on 'Time' column.

# Find the start and end times from expanded_data
sleep_start_time = expanded_data['Time'].min()
sleep_end_time = expanded_data['Time'].max()

# Filter master_data to include only rows within the expanded_data start and end times
master_data = master_data[(master_data['Time'] >= sleep_start_time) & (master_data['Time'] <= sleep_end_time)]

# Map values in the 'Predicted_State' column using label_mapping
label_mapping = {'Awake': 0, 'Deep': 1, 'LS': 2, 'REM': 3}
master_data['Value'] = master_data['Value'].map(label_mapping)
#master_data.to_csv("master_data_5_jan.csv")
master_data
```

```python
def safe_eval(x):
    try:
        return eval(x) if isinstance(x, str) else x
    except:
        return x

master_data['Channel1'] = master_data['Channel1'].apply(safe_eval) # Apply safe_eval to convert string representation of lists to actual lists
master_data['Channel2'] = master_data['Channel2'].apply(safe_eval)
master_data['Channel3'] = master_data['Channel3'].apply(safe_eval)
master_data['Channel4'] = master_data['Channel4'].apply(safe_eval)

# Create a new DataFrame with the lists as columns
X = pd.DataFrame({
    'Channel1': master_data['Channel1'].tolist(),
    'Channel2': master_data['Channel2'].tolist(),
    'Channel3': master_data['Channel3'].tolist(),
    'Channel4': master_data['Channel4'].tolist()
})

X = pd.concat([X[col].apply(pd.Series) for col in X.columns], axis=1) # Flatten the lists (assuming each list has the same length)
X.columns = range(X.shape[1]) # Ensure unique column names
y = master_data['Value'].apply(safe_eval).apply(pd.Series) # Flatten the y values

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=5)

xgb_clf = XGBClassifier()
xgb_clf.fit(X_train, y_train)
y_pred = xgb_clf.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
print(f'Accuracy: {accuracy:.2f}')
print(classification_report(y_test.values.flatten(), y_pred))
```
